```python
archivo = next(iter(uploaded))

df = pd.read_csv(archivo)

df.head()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

df["day_num"] = (
    df["Date"] - pd.Timestamp("2022-01-01")
).dt.days + 1

df[["Date", "day_num"]].head()
```

```python
daily = df.groupby(
    "Date",
    as_index=False
)["Daily Max NO2 Concentration"].mean()

daily["day_num"] = (
    daily["Date"] - pd.Timestamp("2022-01-01")
).dt.days + 1

daily.head()
```

```python
x = daily["day_num"].to_numpy()
y = daily["Daily Max NO2 Concentration"].to_numpy()

m, b = np.polyfit(x, y, 1)

y_pred = m*x + b

r2 = 1 - np.sum((y-y_pred)**2) / np.sum((y-y.mean())**2)

print("Pendiente (m):", m)
print("Intercepto (b):", b)
print("R²:", r2)
```

```python
print(f"Ecuación de regresión:")
print(f"y = {m:.6f}x + {b:.6f}")

print(f"\nCoeficiente de determinación:")
print(f"R² = {r2:.6f}")
```

```python
plt.figure(figsize=(11, 5.5))

plt.scatter(
    daily["Date"],
    daily["Daily Max NO2 Concentration"],
    s=18,
    alpha=0.65,
    label="Promedio diario"
)

plt.plot(
    daily["Date"],
    y_pred,
    linewidth=2,
    label=f"Regresión lineal (R² = {r2:.4f})"
)

plt.title("NO₂ en District of Columbia — 2022")
plt.xlabel("Fecha")
plt.ylabel("NO₂ máximo diario promedio (ppb)")

plt.grid(alpha=0.25)
plt.legend()

plt.show()
```
